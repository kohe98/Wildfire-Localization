```python
"""
extract images from video file
"""

import cv2
import os

def extract_frames(video_path, output_folder):
    # Create the output directory if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # Initialize the video capture object
    video_capture = cv2.VideoCapture(video_path)

    frame_count = 0
    success = True

    print(f"Starting extraction from: {video_path}")

    while success:
        # Read the next frame
        success, frame = video_capture.read()

        if success:
            # Construct the filename (e.g., frame_0001.jpg)
            file_name = f"frame_{frame_count:04d}.jpg"
            file_path = os.path.join(output_folder, file_name)

            # Save the frame as a JPEG file
            cv2.imwrite(file_path, frame)

            frame_count += 1
            if frame_count % 100 == 0:
                print(f"Extracted {frame_count} frames...")

    # Release the video object
    video_capture.release()
    print(f"Done! Total frames extracted: {frame_count}")

# Usage
extract_frames('data/siocomm_a_070223_Arcata1_ChantryFire.mp4', 'extracted_frames')
```

```python
import easyocr
import os
import csv
import re

# Initialize the reader (only do this once)
reader = easyocr.Reader(['en'])

def extract_with_easyocr(image_folder, output_csv):
    results = []
    images = sorted([f for f in os.listdir(image_folder) if f.lower().endswith(('.jpg', '.png'))])

    for img_name in images:
        img_path = os.path.join(image_folder, img_name)

        # EasyOCR can take the path directly or a cropped image array
        # It's smart enough to find the text, but cropping still helps speed
        result = reader.readtext(img_path, detail=0) # detail=0 returns just the text strings

        # Combine list into one string to run our regex
        full_text = " ".join(result)

        # Re-use our previous regex logic
        name_match = re.search(r"^(.*?)\s*elev:", full_text)
        elev = re.search(r"elev:(\d+)", full_text)
        x = re.search(r"X:([+-]?\d+\.\d+)", full_text)
        y = re.search(r"Y:([+-]?\d+\.\d+)", full_text)
        z = re.search(r"Z:([+-]?\d+\.\d+)", full_text)

        results.append({
            "camera_name": name_match.group(1).strip() if name_match else "N/A",
            "elev": elev.group(1) if elev else "N/A",
            "x": x.group(1) if x else "N/A",
            "y": y.group(1) if y else "N/A",
            "z": z.group(1) if z else "N/A"
        })
        print(f"Processed {img_name}")

    # Save to CSV
    headers = ['camera_name', 'elev', 'x', 'y', 'z']
    with open(output_csv, 'w', newline='') as f:
        writer = csv.DictWriter(f, fieldnames=headers)
        writer.writeheader()
        writer.writerows(results)

extract_with_easyocr('data/frames', 'camera_parameters.csv')
```
